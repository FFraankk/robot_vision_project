```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
from preprocessing import *
from utils import *
```

```python
CODE_DIR = Path.cwd()
test_img_path = CODE_DIR / "Test_Image_2026.bmp" 
test_img = cv2.imread(str(test_img_path))
print(test_img.shape)
```

```python
test_img_gray = cv2.cvtColor(test_img, cv2.COLOR_BGR2GRAY)
plot_img_hist(test_img_gray, title="Original Image and Histogram")
```

```python
a = 50
b = 170
alpha = 255 / (b - a)
beta = -alpha * a

stretched_image = np.clip(alpha * test_img_gray + beta, 0, 255).astype(np.uint8)
plot_img_hist(stretched_image, title="Stretched Image and Histogram")
```
